Feature Matrix {X}: Columns tenure_months, monthly_charges, support_calls and contract_type represent input features describing customer behavior.

Target Matrix {Y}: Column churn is the binary label (1=churned, 0=retained). Formulating {Y} is the core step of Supervised Learning.



```python
np.random.seed(42)
n_samples = 1000

data = {
    "tenure_months": np.random.randint(1, 72, n_samples),
    "monthly_charges": np.random.uniform(20.0, 120.0, n_samples),
    "support_calls": np.random.poisson(lam=2, size=n_samples),
    "contract_type": np.random.choice(["Month-to-month", "One year", "Two year"], size=n_samples)
}

df = pd.DataFrame(data)

churn_logit= -0.05 * df['tenure_months']+ 0.4 * df ['support_calls'] - 1.0
churn_prob= 1/ (1+np.exp(-churn_logit))
df['churn']=(churn_prob>np.random.uniform(0, 1, n_samples)).astype(int)

print("---Generated Dataset Sample(Problem Frame: Feature Matrix X + Target y)---")
print(df.head())

print ("---Class Balance Inspection (y distribution)---")
print (df['churn'].value_counts(normalize=True))
```

Class Imbalance (82.7% vs 17.3%): It shows why standard accuracy is misleading. If a dummy model predicts 0 for everyone, it achieves 82.7% accuracy while detecting zero churners. Therefore, we need F1 Score or ROC-AUC is necessary here.
